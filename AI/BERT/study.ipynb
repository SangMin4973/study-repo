## BERT
- BERT는 트랜스포머로 구현되었고, 위키피디아와 BooksCorpus와 같은 레이블이 없는 텍스트 데이터로 사전 훈련된 모델이다.
- BERT는 레이블 없는 방대한 데이터로 훈련되어 다른 작업에서 추가 훈련과 함께 하이퍼파라미터 재조정을 통해 성능을 높였다.
- 하이퍼파라미터 재조정을 파인튜닝(Fine-tuning)이라고 한다.


#### BERT의 크기
- BERT는 트랜스포머 인코더를 쌓아올린 구조이다.
- Base 버전에서는 12개, Large버전에서는 24개를 쌓았다.
    - BERT-Base: L=12, D=768, A=12: 110M개의 파라미터
    - BERT-Large: L=24, D=1024, A=16: 340M개의 파라미터
