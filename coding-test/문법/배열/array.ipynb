#### 배열을 선택할 때 고려할 점
- 할당할 수 있는 메모리 크기를 확인해야 함
- 중간에 데이터 삽입이 많은지 확인해야 함
    - 배열은 중간에 삽입하는 작업이 많을수록 시간 복잡도가 높아짐

```python
# 리스트 끝에 데이터 추가
my_list = [1,2,3]
my_list.append(4)
my_list
```

```python
# +연산자로 데이터 추가
my_list = my_list + [5,6]
my_list
```

```python
# insert()로 데이터 삽입
my_list.insert(2, 9999)
my_list
```

```python
# pop()으로 특정 위치 데이터 삭제
popped_element = my_list.pop(2)
my_list, popped_element
```

```python
# remove로 데이터 삭제
my_list.remove(5)
my_list
```

```python
numbers = [1,2,3,4,5]
squares = [num*2 for num in numbers]
numbers, squares
```
