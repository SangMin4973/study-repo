## 해시
- 해시 함수를 사용해서 변환 값을 인덱스로 삼아 키와 값을 저장해서 빠른 데이터 탐색을 제공하는 자료구조
- 해시는 키와 데이터를 일대일 대응하여 저장
    - 데이터에 바로 접근
- 효율적인 데이터 저장과 탐색

#### 해시의 특징
- 해시는 단방향 동작(키->값 찾기 가능, 값->키 찾기 불가능)
- 키값이 인덱스가 되기 때문에 탐색 과정 없음
- 해시 함수를 통해 데이터->키 변환 과정 필요
- 해시 테이블: 키와 값이 저장되어 있는 공간, 딕셔너리 구조
- 버킷: 해시 테이블의 데이터
- 탐색 횟수가 많은 경우 해시 고려

#### 해시 함수
- 데이터를 인덱스로 표현하기 때문에 (데이터 개수 < 해시 테이블 크기) 
- 충돌을 피해야 함(충돌: 2개 이상의 데이터가 같은 키값을 갖는 경우)
- 해시 함수
    - 나눗셈법
        - 키를 소수로 나눈 나머지를 인덱스로 활용하는 방법
        - $h(x) = x mod\ k$
        - 단점: 상황에 따라 아주 큰 소수가 필요(소수는 직접 구해야 함)
    
    - 곱셈법
        - $h(x) = (((x * A) mod\ 1) * m)$
        - m은 최대 버킷 수, A는 황금비(0.6183)
        - mod 1은 소수 부분만 취함
        - 소수 부분에 m(최대 버킷 수)를 곱해 정수 부분을 인덱스로 취함
        - 1 이하의 값에 m을 곱하기때문에 0~(m-1) 범위를 가짐
    
    - 문자열 해싱
        - $hash(s) = (s[0] + s[1]*p + s[2]*p^2 .... s[n-1]*p^{n-1}) mod\ m$
        - p는 31, m은 해시 테이블 최대 크기
        - m으로 나눈 나머지를 채택
        - 단점: 해시 테이블의 크기가 너무 클 수 있음
        - $(a + b) \% c = (a \% c + b \% c) \% c$로 변환


#### 충돌 처리
- 다른 키에 대해서 해시 함수의 결과값(해시 테이블 인덱스)의 값이 같으면 충돌이 일어남

- 체이닝
    - 충돌이 일어나면 해당 버킷에 링크드리스트로 같은 해시값을 가지는 데이터를 연결
        - 링크드리스트: 데이터 요소들을 연결하여 구성한 선형 데이터 구조
    - 단점
        - 해시 테이블 공간 활용성 떨어짐
            - 링크드리스트가 길어지면 해시 테이블 공간을 덜 활용함
        - 검색 성능이 떨어짐
            - 링크드리스트가 길어지면 링크드리스트를 검색할 때 맨 앞부터 검색해 해시 테이블의 장점 사라짐

- 개방 주소법
    - 충돌이 일어나면 빈 버킷을 찾아 충돌값을 삽입
    - 선형 탐사
        - 빈 버킷을 찾을 때 일정한 간격으로 이동하며 탐사
        - $h(k,i) = (h(k) + i) mod\ m$
        - 단점: 충돌이 발생한 값들이 한 버킷에 모이는 군집(클러스터)가 형성됨
    
    - 이중 해싱 방식
        - 해시 함수를 2개 사용
        - $h(k,i) = (h_1(k) + i * h_2(k)) mod\ m$
