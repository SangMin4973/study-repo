## 배열 정렬하기
- 정수 배열을 오름차순 정렬해서 반환하는 solution() 함수를 완성하라.

#### 제약 조건
- 정수 배열의 길이는 2 이상 $10^5$ 이하입니다.
- 정수 배열의 각 데이터 값은 -100,000 이상 100,000 이하입니다.

#### 입출력의 예
입력|출력
-|-
[1,5,2,4,3]|[-5,1,2,3,4]
[2,1,1,3,2,5,4]|[1,1,2,2,3,4,5]
[6,1,7]|[1,6,7]

```python
def solution(arr):
    return arr.sort() # 원본 arr도 변환
#   원본 리스트 유지
#   sorted_arr = sorted(arr)
#   return sorted_arr
```
