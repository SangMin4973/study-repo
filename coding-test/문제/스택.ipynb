## 괄호 짝 맞추기
- 소괄호 '('와 ')'로 구성된 문자열 s가 주어진다.
- 열린 괄호는 자신과 가장 가까운 닫힌 괄호를 만나면 상쇄되며, 이때 반드시 열린 괄호가 닫힌 괄호보다 먼저 와야 한다.
- 이런 식으로 괄호를 없애는 과정을 반복했을 때, 남아있는 괄호가 없으면 짝이 맞다고 할 수 있다.
- 이 때 괄호의 짝이 맞는지 판별하는 solution()함수를 구현하시오.

#### 제약 조건
- s의 길이는 최대 10만이며, 빈 문자열인 경우는 없습니다.
- 괄호의 짝이 맞으면 True, 짝이 맞지 않으면 False를 반환하세요

#### 입출력의 예
s|반환값
-|-
(())()|True
((())()|False

```python
# 문자열s를 리스트로 저장
# 빈 리스트 result 생성
# result에 문자를 하나씩 저장
# ')'가 오면,
# result의 마지막이 '('이면 
# result.pop()
# result의 마지막이 ')'이면
# result에 저장
# 마지막에 result가 빈 배열이면 True반환
# 아니라면 False 반환

def solution(s):
    s = list(s)
    result = []

    for item in s:
        if item == '(':
            result.append(item)
        else:
            if result[-1] == '(':
                result.pop()
            else:
                result.append(item)
    if result == []:
        return True
    else:
        return False
```

```python
s = '((())()'
solution(s)
```

## 10진수를 2진수로 변환하기
- 10진수 decimal을 입력받아 2진수로 변환하고 이를 문자열로 반환하는 solution()함수를 구현하시오.

#### 제약 조건
- decimal은 1000만 이하의 양의 정수입니다.

#### 입출력의 예
decimal|반환값
-|-
10|"1010"
27|"11011"
12345|"11000000111001"

```python
def solution(decimal):
    result = []
    while True:
        quotient = decimal // 2
        remainder = decimal % 2
        result.append(str(remainder))
        decimal = quotient
        if quotient==0 and remainder==1:
            break
    ans = ""
    while result:
        ans += result.pop()
    return ans
```

```python
decimal = 12345
solution(decimal)
```

## 괄호 회전하기
- 다음 규칙을 지키는 문자열을 올바른 괄호 문자열이라고 정의합니다.
    - "()", "[]", "{}"는 올바른 괄호 문자열입니다.
    - 만약 A가 올바른 괄호 문자열이라면 "(A)", "[A]", "{A}"도 올바른 괄호 문자열입니다. 예를 들어 "[]"가 올바른 괄호 문자열이므로, "([])"도 올바른 괄호 문자열입니다.
- 대괄호, 중괄호, 소괄호로 이루어진 문자열 s가 매개변수로 주어집니다. 이 s를 왼쪽으로 $x (0 \leq x < (s의 길이))$ 칸만큼 회전시켰을 때 s가 올바른 괄호 문자열이 되게 하는 x의 개수를 반환하는 solution 함수를 완성하시오.

#### 제약 조건
- s이 길이는 1이상 1000이하 입니다.

#### 입출력의 예
s|result
-|-
"[]\()\{}"|3
"}]()[{"|2

```python
def solution(s):
    s = list(s)
    cnt = 0
    for i in range(len(s)):
        pairs = []
        for j in range(len(s)):
            c = s[(i+j) % len(s)]
            if c in ['(', '{', '[']:
                pairs.append(c)
            else:
                # 닫는 괄호일때, pairs가 비어있다면 break
                if not pairs:
                    break

                if c == ")" and pairs[-1] == "(":
                    pairs.pop()
                elif c == "}" and pairs[-1] == "{":
                    pairs.pop()
                elif c == "]" and pairs[-1] == "[":
                    pairs.pop()
                else:
                    break
        else:
            if not pairs:
                cnt += 1
    return cnt
```
