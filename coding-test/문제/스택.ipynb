## 괄호 짝 맞추기
- 소괄호 '('와 ')'로 구성된 문자열 s가 주어진다.
- 열린 괄호는 자신과 가장 가까운 닫힌 괄호를 만나면 상쇄되며, 이때 반드시 열린 괄호가 닫힌 괄호보다 먼저 와야 한다.
- 이런 식으로 괄호를 없애는 과정을 반복했을 때, 남아있는 괄호가 없으면 짝이 맞다고 할 수 있다.
- 이 때 괄호의 짝이 맞는지 판별하는 solution()함수를 구현하시오.

#### 제약 조건
- s의 길이는 최대 10만이며, 빈 문자열인 경우는 없습니다.
- 괄호의 짝이 맞으면 True, 짝이 맞지 않으면 False를 반환하세요

#### 입출력의 예
s|반환값
-|-
(())()|True
((())()|False

```python
# 문자열s를 리스트로 저장
# 빈 리스트 result 생성
# result에 문자를 하나씩 저장
# ')'가 오면,
# result의 마지막이 '('이면 
# result.pop()
# result의 마지막이 ')'이면
# result에 저장
# 마지막에 result가 빈 배열이면 True반환
# 아니라면 False 반환

def solution(s):
    s = list(s)
    result = []

    for item in s:
        if item == '(':
            result.append(item)
        else:
            if result[-1] == '(':
                result.pop()
            else:
                result.append(item)
    if result == []:
        return True
    else:
        return False
```

```python
s = '((())()'
solution(s)
```

## 10진수를 2진수로 변환하기
- 10진수 decimal을 입력받아 2진수로 변환하고 이를 문자열로 반환하는 solution()함수를 구현하시오.

#### 제약 조건
- decimal은 1000만 이하의 양의 정수입니다.

#### 입출력의 예
decimal|반환값
-|-
10|"1010"
27|"11011"
12345|"11000000111001"

```python
def solution(decimal):
    result = []
    while True:
        quotient = decimal // 2
        remainder = decimal % 2
        result.append(str(remainder))
        decimal = quotient
        if quotient==0 and remainder==1:
            break
    ans = ""
    while result:
        ans += result.pop()
    return ans
```

## 괄호 회전하기
- 다음 규칙을 지키는 문자열을 올바른 괄호 문자열이라고 정의합니다.
    - "()", "[]", "{}"는 올바른 괄호 문자열입니다.
    - 만약 A가 올바른 괄호 문자열이라면 "(A)", "[A]", "{A}"도 올바른 괄호 문자열입니다. 예를 들어 "[]"가 올바른 괄호 문자열이므로, "([])"도 올바른 괄호 문자열입니다.
- 대괄호, 중괄호, 소괄호로 이루어진 문자열 s가 매개변수로 주어집니다. 이 s를 왼쪽으로 $x (0 \leq x < (s의 길이))$ 칸만큼 회전시켰을 때 s가 올바른 괄호 문자열이 되게 하는 x의 개수를 반환하는 solution 함수를 완성하시오.

#### 제약 조건
- s이 길이는 1이상 1000이하 입니다.

#### 입출력의 예
s|result
-|-
"[]\()\{}"|3
"}]()[{"|2

```python
def solution(s):
    s = list(s)
    cnt = 0
    for i in range(len(s)):
        pairs = []
        for j in range(len(s)):
            c = s[(i+j) % len(s)]
            if c in ['(', '{', '[']:
                pairs.append(c)
            else:
                # 닫는 괄호일때, pairs가 비어있다면 break
                if not pairs:
                    break

                if c == ")" and pairs[-1] == "(":
                    pairs.pop()
                elif c == "}" and pairs[-1] == "{":
                    pairs.pop()
                elif c == "]" and pairs[-1] == "[":
                    pairs.pop()
                else:
                    break
        else:
            if not pairs:
                cnt += 1
    return cnt
```

## 주식 가격
- 초 단위로 기록된 주식 가격이 담긴 배열 prices가 매개변수로 주어질 때, 가격이 떨어지지 않은 기간은 몇 초인지를 반환하도록 solution()함수를 완성하시오.

#### 제약 조건
- pries의 각 가격은 1 이상 10,000 이하인 자연수입니다.
- prices의 길이는 2 이상 100,000 이하입니다.

#### 입출력의 예
prices|return
-|-
[1,2,3,2,3]|[4,3,1,1,0]

```python
# O(N^2)이므로 1000000개의 prices 연산 불가능
def solution(prices):
    result = [0] * len(prices)

    for i in range(1, len(prices)):
        for j in range(i, len(prices)):
            if i == 1 and j == 1:
                result[i-1] += 1
            else:
                if prices[i-1] <= prices[j]:
                    result[i-1] += 1
    
    return result
```

## 크레인 인형 뽑기 게임
- 게임 화면은 1 $\times$ 1 크기의 격자로 구성된 N $\times$ N 크기의 격자이며 위쪽에는 크레인이 있고 오른쪽에는 바구니가 있다.
- 각 격자 칸 안에는 다양한 인형이 있고 인형이 없는 칸은 빈칸이다.
- 인형은 격자 한 칸을 차지하며 격자의 가장 아래 칸부터 차곡차곡 쌓인다.
- 플레이어는 크레인을 좌우로 움직일 수 있고 크레인을 멈춘 위치에서 가장 위에 있는 인형을 집어 올릴 수 있다.
- 집어 올린 인형은 바구니에 쌓인다.
- 이 때 바구니의 가장 아래 칸부터 인형이 순서대로 쌓인다.
- 바구니에 같은 모양의 인형 2개가 연속으로 쌓이면 두 인형은 사라진다.
- 만약 인형이 없는 곳에서 크레인을 작동시키면 아무 일도 일어나지 않는다.
- 바구니는 모든 인형이 들어갈만큼 충분히 크다.
- 2차원 배열 board와 인형을 집는 크레인을 작동시킨 위치가 담긴 배열 moves가 주어질 때, 크레인을 모두 작동시킨 후 사라진 인형 개수를 반환하는 solution()함수를 완성하시오.

#### 제약 조건
- board는 2차원 배열, 크기는 5 $\times$ 5 이상 30 $\times$ 30 이하이다.
- board의 각 칸에는 0 이상 100 이하인 정수가 담겨있다.
    - 0은 빈칸을 의미
    - 1 ~ 100의 각 숫자가 각기 다른인형의 모양을 의미하며 같은 숫자는 같은 모양의 인형을 나타낸다.
- moves배열 크기는 1이상 100 이하이다.
- moves 배열 가 원소들의 값은 1 이상이며 board 배열의 가로 크기 이하인 자연수이다.

#### 입출력의 예
board|moves|result
-|-|-
[[0,0,0,0,0],[0,0,1,0,3],[0,2,5,0,1],[4,2,4,4,2],[3,5,1,3,1]]|[1,5,3,5,1,2,1,4]|4

```python
def solution(board, moves):
    items = [[0]*len(board) for _ in range(len(board[0]))]
    for i in range(len(board)):
        for j in range(len(board[0])):
            items[j][i] = board[i][j]

    bucket = []
    cnt = 0
    for move in moves:
        move -= 1
        for i in range(len(items[move])):
            if items[move][i]!= 0:
                item = items[move][i]
                items[move][i] = 0
                print(item)
                break
            else:
                continue  
        if bucket and bucket[-1] == item:
            bucket.pop()
            cnt += 1
        else:
            bucket.append(item)
             

    return cnt
```
